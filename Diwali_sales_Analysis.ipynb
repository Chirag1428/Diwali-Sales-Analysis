```python
count=sns.countplot(x='Gender',data=df)
for bars in count.containers:
  count.bar_label(bars)
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sales_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data= sales_gen)
```

From the above graph, we conclude that most of the buyers are female and even the purchasing power of females are greater than male.

```python
Age=sns.countplot(data=df,x='Age Group',hue='Gender')
for bars in Age.containers:
  Age.bar_label(bars)
```

```python
Sales_age=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Age Group',y='Amount',data=Sales_age)
```

From the above graph, we conclude that the most of the buyers are age group between 26-35 in female category.

```python
Sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(25,10)})
sns.barplot(data=Sales_state,x='State',y='Orders')
```

```python
Sales_state=df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(25,10)})
sns.barplot(data=Sales_state,x='State',y='Amount')
```

From the above graph, we conclude that most of the orders are from the states Uttar Pradesh, Maharastra and Karnataka respectively.

```python
Marital_status=sns.countplot(data=df,x='Marital_Status')
sns.set(rc={'figure.figsize':(10,5)})
for bars in Marital_status.containers:
  Marital_status.bar_label(bars)
```

```python
Sales_state=df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.set(rc={'figure.figsize':(10,5)})
sns.barplot(data=Sales_state,x='Marital_Status',y='Amount',hue='Gender')
```

From the above graph, we can conclude that the most of the buyers are married(women) and they have high purchasing power.

```python
Occupation=sns.countplot(data=df,x='Occupation')
sns.set(rc={'figure.figsize':(35,10)})
for bars in Occupation.containers:
  Occupation.bar_label(bars)
```

```python
Sales_occupation=df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)
sns.set(rc={'figure.figsize':(35,20)})
sns.barplot(data=Sales_occupation,x='Occupation',y='Amount')
```

From the above graph, we can conclude that most of the buyers are from IT sector, Healtcenter and Aviation respectively.

```python
Product=sns.countplot(data=df,x='Product_Category')
sns.set(rc={'figure.figsize':(20,10)})
for bars in Product.containers:
  Product.bar_label(bars)
```

```python
Sales_product=df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending= False).head(10)
sns.set(rc={'figure.figsize':(30,20)})
sns.barplot(data=Sales_product,x='Product_Category',y='Amount')
```

From the above graph, we can conclude that the most of the sold products are Food, Clothing and Electronics respectively.

```python
Sales_productid=df.groupby(['Product_ID'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(30,20)})
sns.barplot(data=Sales_productid,x='Product_ID',y='Orders')
```

From the above graph, we can specify the most selling Product_id respectively.
